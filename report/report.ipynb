## Build evaluation report

```python
import pandas as pd

# Configure report source
REPORT = 'ablation_no_penalty'
CSV_PATH = f"{REPORT}.csv"
BASE_MODEL = "robust"
METRICS = [
    "ndcg@5",
    "precision@5",
    "recall@5",
    "coverage@5",
    "ndcg@10",
    "precision@10",
    "recall@10",
    "coverage@10",
    "ndcg@20",
    "precision@20",
    "recall@20",
    "coverage@20",
]

# Load data
df = pd.read_csv(CSV_PATH)
df.sort_values(["split", "model"], inplace=True)

# Report by split + model
# ============================================================
grouped = (
    df.groupby(["split", "model"])[METRICS]
    .agg(["mean", "std"])
)

# Create formatted table: mean ± std
formatted_rows = []

for (split, model), row in grouped.iterrows():

    result = {
        "split": split,
        "model": model,
    }

    for metric in METRICS:
        mean_val = row[(metric, "mean")]
        std_val = row[(metric, "std")]

        result[metric] = f"{mean_val:.4f} ± {std_val:.4f}"

    formatted_rows.append(result)

report_df = pd.DataFrame(formatted_rows)

print("\n=== REPORT BY SPLIT + MODEL ===")
print(report_df)

# Save if needed
report_df.to_csv(f"{REPORT}_by_split.csv", index=False)

# Aggregated report by model
# ============================================================
overall_grouped = (
    df.groupby("model")[METRICS]
    .agg(["mean", "std"])
)

overall_rows = []

for model, row in overall_grouped.iterrows():

    result = {
        "model": model,
    }

    for metric in METRICS:
        mean_val = row[(metric, "mean")]
        std_val = row[(metric, "std")]

        result[metric] = f"{mean_val:.4f} ± {std_val:.4f}"

    overall_rows.append(result)

overall_report_df = pd.DataFrame(overall_rows)

print("\n=== OVERALL REPORT BY MODEL ===")
print(overall_report_df)

overall_report_df.to_csv(f"{REPORT}_overall_report.csv", index=False)
```
